# 08 Model Training

Temporal training, validation and testing using Random Forest, XGBoost, LightGBM and CatBoost.

```python
# ======================================================================
# 08_Model_Training.ipynb
# RANDOM FOREST + XGBOOST + LIGHTGBM + CATBOOST
# TEMPORAL TRAIN / VALIDATION / TEST PIPELINE
# ======================================================================
#
# Input:
#   data/processed/model_data/feature_selection/
#       experiment_B_hierarchical_imputation.csv
#       experiment_C_reduced_features.csv
#       experiment_D_without_GSMaP_Gauge.csv
#       experiment_E_correlation_reduced.csv
#
# Temporal split:
#   Train      : 2017-2020
#   Validation : 2021
#   Test       : 2022
#
# Output:
#   data/processed/model_results/
#       model_comparison_validation.csv
#       model_comparison_test.csv
#       best_model_summary.csv
#       models/*.joblib
#       predictions/*.csv
#       feature_importance/*.csv
#
# Figures:
#   figures/model_training/
# ======================================================================


# ======================================================================
# 1. IMPORT LIBRARIES
# ======================================================================

from pathlib import Path
import json
import time
import warnings

import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.ensemble import RandomForestRegressor
from sklearn.inspection import permutation_importance
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score,
)

warnings.filterwarnings("ignore")


# ======================================================================
# 2. OPTIONAL MODEL IMPORTS
# ======================================================================

MODEL_AVAILABILITY = {}


try:
    from xgboost import XGBRegressor
    MODEL_AVAILABILITY["XGBoost"] = True

except Exception as error:
    XGBRegressor = None
    MODEL_AVAILABILITY["XGBoost"] = False
    print("XGBoost unavailable:", error)


try:
    from lightgbm import LGBMRegressor
    MODEL_AVAILABILITY["LightGBM"] = True

except Exception as error:
    LGBMRegressor = None
    MODEL_AVAILABILITY["LightGBM"] = False
    print("LightGBM unavailable:", error)


try:
    from catboost import CatBoostRegressor
    MODEL_AVAILABILITY["CatBoost"] = True

except Exception as error:
    CatBoostRegressor = None
    MODEL_AVAILABILITY["CatBoost"] = False
    print("CatBoost unavailable:", error)


# ======================================================================
# 3. PROJECT CONFIGURATION
# ======================================================================

PROJECT_ROOT = Path(
    r"E:\Geospatial\Precipitation Downscaling"
    r"\Precipitation-Downscaling-Khulna"
)


FEATURE_SELECTION_FOLDER = (
    PROJECT_ROOT
    / "data"
    / "processed"
    / "model_data"
    / "feature_selection"
)


LOG_FOLDER = (
    PROJECT_ROOT
    / "data"
    / "processed"
    / "logs"
)


RESULT_FOLDER = (
    PROJECT_ROOT
    / "data"
    / "processed"
    / "model_results"
)


MODEL_FOLDER = (
    RESULT_FOLDER
    / "models"
)


PREDICTION_FOLDER = (
    RESULT_FOLDER
    / "predictions"
)


IMPORTANCE_FOLDER = (
    RESULT_FOLDER
    / "feature_importance"
)


FIGURE_FOLDER = (
    PROJECT_ROOT
    / "figures"
    / "model_training"
)


for folder in [
    RESULT_FOLDER,
    MODEL_FOLDER,
    PREDICTION_FOLDER,
    IMPORTANCE_FOLDER,
    FIGURE_FOLDER,
    LOG_FOLDER,
]:
    folder.mkdir(
        parents=True,
        exist_ok=True
    )


# ======================================================================
# 4. SETTINGS
# ======================================================================

TARGET_COLUMN = "rainfall_mm"

RANDOM_STATE = 42

N_JOBS = -1

CLIP_NEGATIVE_PREDICTIONS = True

PERMUTATION_REPEATS = 30


EXPERIMENT_FILES = {
    "B_full_imputed":
        FEATURE_SELECTION_FOLDER
        / "experiment_B_hierarchical_imputation.csv",

    "C_reduced_features":
        FEATURE_SELECTION_FOLDER
        / "experiment_C_reduced_features.csv",

    "D_without_GSMaP_Gauge":
        FEATURE_SELECTION_FOLDER
        / "experiment_D_without_GSMaP_Gauge.csv",

    "E_correlation_reduced":
        FEATURE_SELECTION_FOLDER
        / "experiment_E_correlation_reduced.csv",
}


FEATURE_SET_JSON = (
    LOG_FOLDER
    / "feature_selection_feature_sets.json"
)


NON_FEATURE_COLUMNS = {
    "station_id",
    "station_name",
    "date",
    "month_id",
    "season",
    "split",
    TARGET_COLUMN,
    "raster_x",
    "raster_y",
    "inside_raster_bounds",
    "valid_predictor_count",
    "missing_predictor_count",
    "all_predictors_available",
}


# Compact parameter grids suitable for a small dataset
MODEL_PARAM_GRIDS = {
    "Random_Forest": [
        {
            "n_estimators": 500,
            "max_depth": None,
            "min_samples_leaf": 1,
            "max_features": 1.0,
        },
        {
            "n_estimators": 500,
            "max_depth": 12,
            "min_samples_leaf": 2,
            "max_features": 0.8,
        },
        {
            "n_estimators": 700,
            "max_depth": 8,
            "min_samples_leaf": 3,
            "max_features": "sqrt",
        },
    ],
}


if MODEL_AVAILABILITY.get("XGBoost"):

    MODEL_PARAM_GRIDS["XGBoost"] = [
        {
            "n_estimators": 500,
            "learning_rate": 0.03,
            "max_depth": 3,
            "min_child_weight": 3,
            "subsample": 0.9,
            "colsample_bytree": 0.9,
            "reg_alpha": 0.0,
            "reg_lambda": 1.0,
        },
        {
            "n_estimators": 400,
            "learning_rate": 0.05,
            "max_depth": 2,
            "min_child_weight": 3,
            "subsample": 0.9,
            "colsample_bytree": 0.8,
            "reg_alpha": 0.1,
            "reg_lambda": 3.0,
        },
        {
            "n_estimators": 600,
            "learning_rate": 0.02,
            "max_depth": 4,
            "min_child_weight": 5,
            "subsample": 0.85,
            "colsample_bytree": 0.85,
            "reg_alpha": 0.1,
            "reg_lambda": 5.0,
        },
    ]


if MODEL_AVAILABILITY.get("LightGBM"):

    MODEL_PARAM_GRIDS["LightGBM"] = [
        {
            "n_estimators": 500,
            "learning_rate": 0.03,
            "num_leaves": 15,
            "max_depth": -1,
            "min_child_samples": 15,
            "subsample": 0.9,
            "colsample_bytree": 0.9,
            "reg_alpha": 0.0,
            "reg_lambda": 1.0,
        },
        {
            "n_estimators": 400,
            "learning_rate": 0.05,
            "num_leaves": 10,
            "max_depth": 5,
            "min_child_samples": 20,
            "subsample": 0.9,
            "colsample_bytree": 0.8,
            "reg_alpha": 0.1,
            "reg_lambda": 3.0,
        },
    ]


if MODEL_AVAILABILITY.get("CatBoost"):

    MODEL_PARAM_GRIDS["CatBoost"] = [
        {
            "iterations": 500,
            "learning_rate": 0.03,
            "depth": 4,
            "l2_leaf_reg": 3.0,
        },
        {
            "iterations": 400,
            "learning_rate": 0.05,
            "depth": 5,
            "l2_leaf_reg": 5.0,
        },
        {
            "iterations": 600,
            "learning_rate": 0.02,
            "depth": 6,
            "l2_leaf_reg": 7.0,
        },
    ]


# ======================================================================
# 5. HELPER FUNCTIONS
# ======================================================================

def save_figure(filename):
    """
    Save current matplotlib figure.
    """

    output_path = (
        FIGURE_FOLDER
        / filename
    )

    plt.tight_layout()

    plt.savefig(
        output_path,
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()

    plt.close()

    return output_path


def safe_name(text):
    """
    Create a safe filename.
    """

    return (
        str(text)
        .replace(" ", "_")
        .replace("/", "_")
        .replace("\\", "_")
    )


def load_feature_sets():
    """
    Load feature-set definitions created by Notebook 07.
    """

    if not FEATURE_SET_JSON.exists():
        return {}

    try:

        with open(
            FEATURE_SET_JSON,
            "r",
            encoding="utf-8"
        ) as file:

            return json.load(
                file
            )

    except Exception as error:

        print(
            "WARNING: Feature-set JSON could not be read."
        )

        print(error)

        return {}


def detect_numeric_features(
    dataframe,
    experiment_name,
    feature_sets
):
    """
    Select features from JSON when available.
    Otherwise use numeric columns excluding metadata and QC fields.
    """

    mapping = {
        "B_full_imputed":
            "experiment_B_hierarchical_imputation",

        "C_reduced_features":
            "experiment_C_reduced_features",

        "D_without_GSMaP_Gauge":
            "experiment_D_without_GSMaP_Gauge",

        "E_correlation_reduced":
            "experiment_E_correlation_reduced",
    }


    json_key = mapping.get(
        experiment_name
    )


    if (
        json_key in feature_sets
        and isinstance(
            feature_sets[
                json_key
            ],
            list
        )
    ):

        features = [
            feature
            for feature in feature_sets[
                json_key
            ]
            if feature in dataframe.columns
        ]

    else:

        features = [
            column
            for column in dataframe.columns
            if column not in NON_FEATURE_COLUMNS
            and pd.api.types.is_numeric_dtype(
                dataframe[
                    column
                ]
            )
        ]


    features = list(
        dict.fromkeys(
            features
        )
    )


    if TARGET_COLUMN in features:

        features.remove(
            TARGET_COLUMN
        )


    return features


def validate_dataset(
    dataframe,
    experiment_name,
    features
):
    """
    Validate model-training dataset.
    """

    required_columns = [
        "station_id",
        "year",
        "month",
        "split",
        TARGET_COLUMN,
    ]


    missing_required = [
        column
        for column in required_columns
        if column not in dataframe.columns
    ]


    if missing_required:

        raise ValueError(
            f"{experiment_name}: required columns missing: "
            + ", ".join(
                missing_required
            )
        )


    if not features:

        raise ValueError(
            f"{experiment_name}: no features found."
        )


    duplicate_rows = int(
        dataframe.duplicated(
            subset=[
                "station_id",
                "year",
                "month"
            ]
        ).sum()
    )


    if duplicate_rows > 0:

        raise ValueError(
            f"{experiment_name}: "
            f"{duplicate_rows} duplicate rows."
        )


    missing_target = int(
        dataframe[
            TARGET_COLUMN
        ].isna().sum()
    )


    if missing_target > 0:

        raise ValueError(
            f"{experiment_name}: "
            f"{missing_target} missing target values."
        )


    missing_features = int(
        dataframe[
            features
        ]
        .isna()
        .sum()
        .sum()
    )


    if missing_features > 0:

        raise ValueError(
            f"{experiment_name}: "
            f"{missing_features} missing feature values."
        )


    expected_splits = {
        "train",
        "validation",
        "test",
    }


    actual_splits = set(
        dataframe[
            "split"
        ].dropna().unique()
    )


    missing_splits = (
        expected_splits
        - actual_splits
    )


    if missing_splits:

        raise ValueError(
            f"{experiment_name}: missing split(s): "
            + ", ".join(
                sorted(
                    missing_splits
                )
            )
        )


def build_model(
    model_name,
    parameters
):
    """
    Build selected regression model.
    """

    if model_name == "Random_Forest":

        return RandomForestRegressor(
            random_state=RANDOM_STATE,
            n_jobs=N_JOBS,
            **parameters
        )


    if (
        model_name == "XGBoost"
        and XGBRegressor is not None
    ):

        return XGBRegressor(
            objective="reg:squarederror",
            random_state=RANDOM_STATE,
            n_jobs=N_JOBS,
            verbosity=0,
            **parameters
        )


    if (
        model_name == "LightGBM"
        and LGBMRegressor is not None
    ):

        return LGBMRegressor(
            random_state=RANDOM_STATE,
            n_jobs=N_JOBS,
            verbosity=-1,
            **parameters
        )


    if (
        model_name == "CatBoost"
        and CatBoostRegressor is not None
    ):

        return CatBoostRegressor(
            random_seed=RANDOM_STATE,
            verbose=False,
            allow_writing_files=False,
            **parameters
        )


    raise ValueError(
        f"Unknown or unavailable model: {model_name}"
    )


def calculate_metrics(
    observed,
    predicted
):
    """
    Calculate rainfall-regression evaluation metrics.
    """

    observed = np.asarray(
        observed,
        dtype=float
    )

    predicted = np.asarray(
        predicted,
        dtype=float
    )


    valid_mask = (
        np.isfinite(
            observed
        )
        &
        np.isfinite(
            predicted
        )
    )


    observed = observed[
        valid_mask
    ]

    predicted = predicted[
        valid_mask
    ]


    if len(
        observed
    ) == 0:

        return {
            "N": 0,
            "RMSE": np.nan,
            "MAE": np.nan,
            "R2": np.nan,
            "Bias": np.nan,
            "Percent_Bias": np.nan,
            "Pearson_R": np.nan,
            "NSE": np.nan,
            "KGE": np.nan,
        }


    residual = (
        predicted
        - observed
    )


    rmse = float(
        np.sqrt(
            mean_squared_error(
                observed,
                predicted
            )
        )
    )


    mae = float(
        mean_absolute_error(
            observed,
            predicted
        )
    )


    r2 = float(
        r2_score(
            observed,
            predicted
        )
    )


    bias = float(
        np.mean(
            residual
        )
    )


    observed_sum = float(
        np.sum(
            observed
        )
    )


    percent_bias = (
        float(
            100
            * np.sum(
                residual
            )
            / observed_sum
        )
        if observed_sum != 0
        else np.nan
    )


    if (
        len(
            observed
        ) > 1
        and np.std(
            observed
        ) > 0
        and np.std(
            predicted
        ) > 0
    ):

        pearson_r = float(
            np.corrcoef(
                observed,
                predicted
            )[
                0,
                1
            ]
        )

    else:

        pearson_r = np.nan


    nse_denominator = float(
        np.sum(
            (
                observed
                - np.mean(
                    observed
                )
            )
            ** 2
        )
    )


    nse = (
        float(
            1
            - (
                np.sum(
                    (
                        predicted
                        - observed
                    )
                    ** 2
                )
                / nse_denominator
            )
        )
        if nse_denominator != 0
        else np.nan
    )


    observed_mean = float(
        np.mean(
            observed
        )
    )


    predicted_mean = float(
        np.mean(
            predicted
        )
    )


    observed_std = float(
        np.std(
            observed,
            ddof=1
        )
    ) if len(
        observed
    ) > 1 else np.nan


    predicted_std = float(
        np.std(
            predicted,
            ddof=1
        )
    ) if len(
        predicted
    ) > 1 else np.nan


    alpha = (
        predicted_std
        / observed_std
        if (
            np.isfinite(
                observed_std
            )
            and observed_std != 0
        )
        else np.nan
    )


    beta = (
        predicted_mean
        / observed_mean
        if observed_mean != 0
        else np.nan
    )


    kge = (
        float(
            1
            - np.sqrt(
                (
                    pearson_r
                    - 1
                )
                ** 2
                + (
                    alpha
                    - 1
                )
                ** 2
                + (
                    beta
                    - 1
                )
                ** 2
            )
        )
        if (
            np.isfinite(
                pearson_r
            )
            and np.isfinite(
                alpha
            )
            and np.isfinite(
                beta
            )
        )
        else np.nan
    )


    return {
        "N": int(
            len(
                observed
            )
        ),
        "RMSE": rmse,
        "MAE": mae,
        "R2": r2,
        "Bias": bias,
        "Percent_Bias": percent_bias,
        "Pearson_R": pearson_r,
        "NSE": nse,
        "KGE": kge,
    }


def create_prediction_dataframe(
    source_dataframe,
    observed,
    predicted,
    experiment_name,
    model_name,
    split_name
):
    """
    Create detailed prediction table.
    """

    metadata_columns = [
        column
        for column in [
            "station_id",
            "station_name",
            "latitude",
            "longitude",
            "year",
            "month",
            "date",
            "month_id",
            "season",
            "split",
        ]
        if column in source_dataframe.columns
    ]


    output_dataframe = (
        source_dataframe[
            metadata_columns
        ]
        .copy()
    )


    output_dataframe[
        "Observed_Rainfall_mm"
    ] = np.asarray(
        observed,
        dtype=float
    )


    output_dataframe[
        "Predicted_Rainfall_mm"
    ] = np.asarray(
        predicted,
        dtype=float
    )


    output_dataframe[
        "Residual_mm"
    ] = (
        output_dataframe[
            "Predicted_Rainfall_mm"
        ]
        - output_dataframe[
            "Observed_Rainfall_mm"
        ]
    )


    output_dataframe[
        "Absolute_Error_mm"
    ] = (
        output_dataframe[
            "Residual_mm"
        ].abs()
    )


    output_dataframe[
        "Experiment"
    ] = experiment_name


    output_dataframe[
        "Model"
    ] = model_name


    output_dataframe[
        "Evaluation_Split"
    ] = split_name


    return output_dataframe


def native_feature_importance(
    fitted_model,
    feature_names
):
    """
    Extract model-native feature importance when available.
    """

    if hasattr(
        fitted_model,
        "feature_importances_"
    ):

        values = np.asarray(
            fitted_model.feature_importances_,
            dtype=float
        )


        if len(
            values
        ) == len(
            feature_names
        ):

            return (
                pd.DataFrame({
                    "Feature": feature_names,
                    "Importance": values,
                    "Importance_Type":
                        "Native_Feature_Importance",
                })
                .sort_values(
                    "Importance",
                    ascending=False
                )
                .reset_index(
                    drop=True
                )
            )


    return pd.DataFrame()


def permutation_importance_table(
    fitted_model,
    features,
    target,
    feature_names
):
    """
    Calculate permutation importance on validation/test data.
    """

    try:

        result = permutation_importance(
            estimator=fitted_model,
            X=features,
            y=target,
            scoring="neg_root_mean_squared_error",
            n_repeats=PERMUTATION_REPEATS,
            random_state=RANDOM_STATE,
            n_jobs=N_JOBS,
        )


        return (
            pd.DataFrame({
                "Feature": feature_names,
                "Importance": (
                    result.importances_mean
                ),
                "Importance_Std": (
                    result.importances_std
                ),
                "Importance_Type":
                    "Permutation_RMSE_Decrease",
            })
            .sort_values(
                "Importance",
                ascending=False
            )
            .reset_index(
                drop=True
            )
        )


    except Exception as error:

        print(
            "Permutation importance failed:"
        )

        print(error)

        return pd.DataFrame()


# ======================================================================
# 6. INITIAL VALIDATION
# ======================================================================

print("=" * 88)
print("TEMPORAL MACHINE-LEARNING MODEL TRAINING")
print("=" * 88)

print(
    f"Project root             : "
    f"{PROJECT_ROOT}"
)

print(
    f"Feature selection folder : "
    f"{FEATURE_SELECTION_FOLDER}"
)

print(
    f"Result folder            : "
    f"{RESULT_FOLDER}"
)

print(
    f"Model folder             : "
    f"{MODEL_FOLDER}"
)

print(
    f"Figure folder            : "
    f"{FIGURE_FOLDER}"
)

print()


if not FEATURE_SELECTION_FOLDER.exists():

    raise FileNotFoundError(
        f"Feature-selection folder পাওয়া যায়নি:\n"
        f"{FEATURE_SELECTION_FOLDER}"
    )


print(
    "Available models:"
)

for model_name in [
    "Random_Forest",
    "XGBoost",
    "LightGBM",
    "CatBoost",
]:

    available = (
        model_name in MODEL_PARAM_GRIDS
    )

    print(
        f"  {model_name}: {available}"
    )


feature_sets = load_feature_sets()


# ======================================================================
# 7. LOAD EXPERIMENT DATASETS
# ======================================================================

experiment_data = {}

experiment_features = {}

experiment_inventory_records = []


for experiment_name, file_path in (
    EXPERIMENT_FILES.items()
):

    if not file_path.exists():

        print(
            f"WARNING: Skipping missing experiment: "
            f"{file_path.name}"
        )

        experiment_inventory_records.append({
            "Experiment": experiment_name,
            "File": str(
                file_path
            ),
            "Status": "Missing",
            "Rows": 0,
            "Feature_Count": 0,
        })

        continue


    dataframe = pd.read_csv(
        file_path
    )


    if "date" in dataframe.columns:

        dataframe[
            "date"
        ] = pd.to_datetime(
            dataframe[
                "date"
            ],
            errors="coerce"
        )


    numeric_candidate_columns = [
        column
        for column in dataframe.columns
        if column not in {
            "station_id",
            "station_name",
            "date",
            "month_id",
            "season",
            "split",
        }
    ]


    for column in numeric_candidate_columns:

        dataframe[
            column
        ] = pd.to_numeric(
            dataframe[
                column
            ],
            errors="coerce"
        )


    features = detect_numeric_features(
        dataframe=dataframe,
        experiment_name=experiment_name,
        feature_sets=feature_sets,
    )


    validate_dataset(
        dataframe=dataframe,
        experiment_name=experiment_name,
        features=features,
    )


    experiment_data[
        experiment_name
    ] = dataframe


    experiment_features[
        experiment_name
    ] = features


    experiment_inventory_records.append({
        "Experiment": experiment_name,
        "File": str(
            file_path
        ),
        "Status": "Loaded",
        "Rows": len(
            dataframe
        ),
        "Feature_Count": len(
            features
        ),
        "Train_Rows": int(
            (
                dataframe[
                    "split"
                ]
                == "train"
            ).sum()
        ),
        "Validation_Rows": int(
            (
                dataframe[
                    "split"
                ]
                == "validation"
            ).sum()
        ),
        "Test_Rows": int(
            (
                dataframe[
                    "split"
                ]
                == "test"
            ).sum()
        ),
    })


experiment_inventory_df = (
    pd.DataFrame(
        experiment_inventory_records
    )
)


print()
print("=" * 88)
print("EXPERIMENT INVENTORY")
print("=" * 88)

display(
    experiment_inventory_df
)


if not experiment_data:

    raise RuntimeError(
        "কোনো valid experiment dataset পাওয়া যায়নি।"
    )


# ======================================================================
# 8. VALIDATION-BASED MODEL SELECTION
# ======================================================================

candidate_records = []

best_configurations = {}


for experiment_name, dataframe in (
    experiment_data.items()
):

    features = experiment_features[
        experiment_name
    ]


    train_df = dataframe[
        dataframe[
            "split"
        ]
        == "train"
    ].copy()


    validation_df = dataframe[
        dataframe[
            "split"
        ]
        == "validation"
    ].copy()


    X_train = (
        train_df[
            features
        ]
        .astype(float)
    )


    y_train = (
        train_df[
            TARGET_COLUMN
        ]
        .astype(float)
    )


    X_validation = (
        validation_df[
            features
        ]
        .astype(float)
    )


    y_validation = (
        validation_df[
            TARGET_COLUMN
        ]
        .astype(float)
    )


    print()
    print("=" * 88)
    print(
        f"EXPERIMENT: {experiment_name}"
    )
    print("=" * 88)

    print(
        f"Train rows      : "
        f"{len(train_df)}"
    )

    print(
        f"Validation rows : "
        f"{len(validation_df)}"
    )

    print(
        f"Feature count   : "
        f"{len(features)}"
    )


    for model_name, parameter_grid in (
        MODEL_PARAM_GRIDS.items()
    ):

        best_model_record = None


        for candidate_number, parameters in enumerate(
            parameter_grid,
            start=1
        ):

            start_time = time.time()


            try:

                model = build_model(
                    model_name=model_name,
                    parameters=parameters,
                )


                model.fit(
                    X_train,
                    y_train
                )


                train_prediction = (
                    model.predict(
                        X_train
                    )
                )


                validation_prediction = (
                    model.predict(
                        X_validation
                    )
                )


                if CLIP_NEGATIVE_PREDICTIONS:

                    train_prediction = np.clip(
                        train_prediction,
                        0,
                        None
                    )


                    validation_prediction = np.clip(
                        validation_prediction,
                        0,
                        None
                    )


                train_metrics = (
                    calculate_metrics(
                        observed=y_train,
                        predicted=train_prediction,
                    )
                )


                validation_metrics = (
                    calculate_metrics(
                        observed=y_validation,
                        predicted=validation_prediction,
                    )
                )


                elapsed_seconds = (
                    time.time()
                    - start_time
                )


                candidate_record = {
                    "Experiment": experiment_name,
                    "Model": model_name,
                    "Candidate_Number":
                        candidate_number,
                    "Parameters": json.dumps(
                        parameters,
                        sort_keys=True
                    ),
                    "Feature_Count": len(
                        features
                    ),
                    "Train_Rows": len(
                        train_df
                    ),
                    "Validation_Rows": len(
                        validation_df
                    ),
                    "Training_Seconds":
                        elapsed_seconds,

                    "Train_RMSE":
                        train_metrics[
                            "RMSE"
                        ],
                    "Train_MAE":
                        train_metrics[
                            "MAE"
                        ],
                    "Train_R2":
                        train_metrics[
                            "R2"
                        ],

                    "Validation_RMSE":
                        validation_metrics[
                            "RMSE"
                        ],
                    "Validation_MAE":
                        validation_metrics[
                            "MAE"
                        ],
                    "Validation_R2":
                        validation_metrics[
                            "R2"
                        ],
                    "Validation_Bias":
                        validation_metrics[
                            "Bias"
                        ],
                    "Validation_Percent_Bias":
                        validation_metrics[
                            "Percent_Bias"
                        ],
                    "Validation_Pearson_R":
                        validation_metrics[
                            "Pearson_R"
                        ],
                    "Validation_NSE":
                        validation_metrics[
                            "NSE"
                        ],
                    "Validation_KGE":
                        validation_metrics[
                            "KGE"
                        ],

                    "Status": "Success",
                    "Error": "",
                }


                candidate_records.append(
                    candidate_record
                )


                if (
                    best_model_record is None
                    or validation_metrics[
                        "RMSE"
                    ]
                    < best_model_record[
                        "Validation_RMSE"
                    ]
                ):

                    best_model_record = {
                        **candidate_record,
                        "Parameters_Dict":
                            parameters,
                    }


                print(
                    f"{model_name:<18} "
                    f"Candidate {candidate_number} | "
                    f"Validation RMSE="
                    f"{validation_metrics['RMSE']:.4f} | "
                    f"R2="
                    f"{validation_metrics['R2']:.4f}"
                )


            except Exception as error:

                candidate_records.append({
                    "Experiment": experiment_name,
                    "Model": model_name,
                    "Candidate_Number":
                        candidate_number,
                    "Parameters": json.dumps(
                        parameters,
                        sort_keys=True
                    ),
                    "Feature_Count": len(
                        features
                    ),
                    "Train_Rows": len(
                        train_df
                    ),
                    "Validation_Rows": len(
                        validation_df
                    ),
                    "Training_Seconds":
                        np.nan,

                    "Train_RMSE":
                        np.nan,
                    "Train_MAE":
                        np.nan,
                    "Train_R2":
                        np.nan,

                    "Validation_RMSE":
                        np.nan,
                    "Validation_MAE":
                        np.nan,
                    "Validation_R2":
                        np.nan,
                    "Validation_Bias":
                        np.nan,
                    "Validation_Percent_Bias":
                        np.nan,
                    "Validation_Pearson_R":
                        np.nan,
                    "Validation_NSE":
                        np.nan,
                    "Validation_KGE":
                        np.nan,

                    "Status": "Failed",
                    "Error": str(
                        error
                    ),
                })


                print(
                    f"FAILED: {experiment_name} | "
                    f"{model_name} | "
                    f"Candidate {candidate_number}"
                )

                print(error)


        if best_model_record is not None:

            best_configurations[
                (
                    experiment_name,
                    model_name
                )
            ] = best_model_record


candidate_results_df = (
    pd.DataFrame(
        candidate_records
    )
)


successful_candidate_df = (
    candidate_results_df[
        candidate_results_df[
            "Status"
        ]
        == "Success"
    ]
    .copy()
)


if successful_candidate_df.empty:

    raise RuntimeError(
        "সব model candidate failed হয়েছে।"
    )


# ======================================================================
# 9. REFIT BEST CONFIGURATIONS AND TEST ON 2022
# ======================================================================

validation_best_records = []

test_result_records = []

all_test_predictions = []

best_model_objects = {}

feature_importance_records = []


for (
    experiment_name,
    model_name
), best_record in (
    best_configurations.items()
):

    dataframe = experiment_data[
        experiment_name
    ]


    features = experiment_features[
        experiment_name
    ]


    train_df = dataframe[
        dataframe[
            "split"
        ]
        == "train"
    ].copy()


    validation_df = dataframe[
        dataframe[
            "split"
        ]
        == "validation"
    ].copy()


    test_df = dataframe[
        dataframe[
            "split"
        ]
        == "test"
    ].copy()


    train_validation_df = pd.concat(
        [
            train_df,
            validation_df,
        ],
        ignore_index=True
    )


    X_train_validation = (
        train_validation_df[
            features
        ]
        .astype(float)
    )


    y_train_validation = (
        train_validation_df[
            TARGET_COLUMN
        ]
        .astype(float)
    )


    X_test = (
        test_df[
            features
        ]
        .astype(float)
    )


    y_test = (
        test_df[
            TARGET_COLUMN
        ]
        .astype(float)
    )


    best_parameters = (
        best_record[
            "Parameters_Dict"
        ]
    )


    final_model = build_model(
        model_name=model_name,
        parameters=best_parameters,
    )


    start_time = time.time()


    final_model.fit(
        X_train_validation,
        y_train_validation
    )


    test_prediction = (
        final_model.predict(
            X_test
        )
    )


    if CLIP_NEGATIVE_PREDICTIONS:

        test_prediction = np.clip(
            test_prediction,
            0,
            None
        )


    test_metrics = calculate_metrics(
        observed=y_test,
        predicted=test_prediction,
    )


    final_training_seconds = (
        time.time()
        - start_time
    )


    model_key = (
        f"{safe_name(experiment_name)}"
        f"__{safe_name(model_name)}"
    )


    model_output = (
        MODEL_FOLDER
        / f"{model_key}.joblib"
    )


    joblib.dump(
        {
            "model": final_model,
            "features": features,
            "target": TARGET_COLUMN,
            "experiment": experiment_name,
            "model_name": model_name,
            "parameters": best_parameters,
            "train_years": [
                2017,
                2018,
                2019,
                2020,
                2021,
            ],
            "test_year": 2022,
            "clip_negative_predictions":
                CLIP_NEGATIVE_PREDICTIONS,
        },
        model_output
    )


    best_model_objects[
        model_key
    ] = final_model


    validation_best_records.append({
        "Experiment": experiment_name,
        "Model": model_name,
        "Parameters": json.dumps(
            best_parameters,
            sort_keys=True
        ),
        "Feature_Count": len(
            features
        ),
        "Validation_RMSE":
            best_record[
                "Validation_RMSE"
            ],
        "Validation_MAE":
            best_record[
                "Validation_MAE"
            ],
        "Validation_R2":
            best_record[
                "Validation_R2"
            ],
        "Validation_Bias":
            best_record[
                "Validation_Bias"
            ],
        "Validation_Percent_Bias":
            best_record[
                "Validation_Percent_Bias"
            ],
        "Validation_Pearson_R":
            best_record[
                "Validation_Pearson_R"
            ],
        "Validation_NSE":
            best_record[
                "Validation_NSE"
            ],
        "Validation_KGE":
            best_record[
                "Validation_KGE"
            ],
    })


    test_result_records.append({
        "Experiment": experiment_name,
        "Model": model_name,
        "Parameters": json.dumps(
            best_parameters,
            sort_keys=True
        ),
        "Feature_Count": len(
            features
        ),
        "Train_Validation_Rows":
            len(
                train_validation_df
            ),
        "Test_Rows": len(
            test_df
        ),
        "Final_Training_Seconds":
            final_training_seconds,
        "Test_RMSE":
            test_metrics[
                "RMSE"
            ],
        "Test_MAE":
            test_metrics[
                "MAE"
            ],
        "Test_R2":
            test_metrics[
                "R2"
            ],
        "Test_Bias":
            test_metrics[
                "Bias"
            ],
        "Test_Percent_Bias":
            test_metrics[
                "Percent_Bias"
            ],
        "Test_Pearson_R":
            test_metrics[
                "Pearson_R"
            ],
        "Test_NSE":
            test_metrics[
                "NSE"
            ],
        "Test_KGE":
            test_metrics[
                "KGE"
            ],
        "Saved_Model": str(
            model_output
        ),
    })


    prediction_df = (
        create_prediction_dataframe(
            source_dataframe=test_df,
            observed=y_test,
            predicted=test_prediction,
            experiment_name=experiment_name,
            model_name=model_name,
            split_name="test",
        )
    )


    prediction_output = (
        PREDICTION_FOLDER
        / f"{model_key}__test_predictions.csv"
    )


    prediction_df.to_csv(
        prediction_output,
        index=False
    )


    all_test_predictions.append(
        prediction_df
    )


    native_importance_df = (
        native_feature_importance(
            fitted_model=final_model,
            feature_names=features,
        )
    )


    if not native_importance_df.empty:

        native_importance_df[
            "Experiment"
        ] = experiment_name

        native_importance_df[
            "Model"
        ] = model_name

        native_importance_df.to_csv(
            IMPORTANCE_FOLDER
            / f"{model_key}__native_importance.csv",
            index=False
        )


        feature_importance_records.append(
            native_importance_df
        )


    permutation_df = (
        permutation_importance_table(
            fitted_model=final_model,
            features=X_test,
            target=y_test,
            feature_names=features,
        )
    )


    if not permutation_df.empty:

        permutation_df[
            "Experiment"
        ] = experiment_name

        permutation_df[
            "Model"
        ] = model_name

        permutation_df.to_csv(
            IMPORTANCE_FOLDER
            / f"{model_key}__permutation_importance.csv",
            index=False
        )


        feature_importance_records.append(
            permutation_df
        )


validation_best_df = pd.DataFrame(
    validation_best_records
)


test_results_df = pd.DataFrame(
    test_result_records
)


all_test_predictions_df = pd.concat(
    all_test_predictions,
    ignore_index=True
)


# ======================================================================
# 10. RANK MODELS
# ======================================================================

validation_best_df = (
    validation_best_df
    .sort_values(
        [
            "Validation_RMSE",
            "Validation_MAE",
        ],
        ascending=[
            True,
            True,
        ]
    )
    .reset_index(
        drop=True
    )
)


validation_best_df[
    "Validation_Rank"
] = np.arange(
    1,
    len(
        validation_best_df
    )
    + 1
)


test_results_df = (
    test_results_df
    .sort_values(
        [
            "Test_RMSE",
            "Test_MAE",
        ],
        ascending=[
            True,
            True,
        ]
    )
    .reset_index(
        drop=True
    )
)


test_results_df[
    "Test_Rank"
] = np.arange(
    1,
    len(
        test_results_df
    )
    + 1
)


best_test_row = (
    test_results_df.iloc[
        0
    ]
)


best_model_summary_df = pd.DataFrame([
    {
        "Best_Experiment":
            best_test_row[
                "Experiment"
            ],
        "Best_Model":
            best_test_row[
                "Model"
            ],
        "Test_RMSE":
            best_test_row[
                "Test_RMSE"
            ],
        "Test_MAE":
            best_test_row[
                "Test_MAE"
            ],
        "Test_R2":
            best_test_row[
                "Test_R2"
            ],
        "Test_Bias":
            best_test_row[
                "Test_Bias"
            ],
        "Test_Percent_Bias":
            best_test_row[
                "Test_Percent_Bias"
            ],
        "Test_Pearson_R":
            best_test_row[
                "Test_Pearson_R"
            ],
        "Test_NSE":
            best_test_row[
                "Test_NSE"
            ],
        "Test_KGE":
            best_test_row[
                "Test_KGE"
            ],
        "Saved_Model":
            best_test_row[
                "Saved_Model"
            ],
    }
])


# ======================================================================
# 11. STATION-WISE AND MONTH-WISE METRICS FOR BEST MODEL
# ======================================================================

best_prediction_df = (
    all_test_predictions_df[
        (
            all_test_predictions_df[
                "Experiment"
            ]
            == best_test_row[
                "Experiment"
            ]
        )
        &
        (
            all_test_predictions_df[
                "Model"
            ]
            == best_test_row[
                "Model"
            ]
        )
    ]
    .copy()
)


station_metric_records = []


for (
    station_id,
    station_name
), station_df in (
    best_prediction_df.groupby(
        [
            "station_id",
            "station_name"
        ]
    )
):

    metrics = calculate_metrics(
        observed=station_df[
            "Observed_Rainfall_mm"
        ],
        predicted=station_df[
            "Predicted_Rainfall_mm"
        ],
    )


    station_metric_records.append({
        "station_id": station_id,
        "station_name": station_name,
        **metrics,
    })


station_metrics_df = pd.DataFrame(
    station_metric_records
)


monthly_metric_records = []


for month, month_df in (
    best_prediction_df.groupby(
        "month"
    )
):

    metrics = calculate_metrics(
        observed=month_df[
            "Observed_Rainfall_mm"
        ],
        predicted=month_df[
            "Predicted_Rainfall_mm"
        ],
    )


    monthly_metric_records.append({
        "month": month,
        **metrics,
    })


monthly_metrics_df = pd.DataFrame(
    monthly_metric_records
)


# ======================================================================
# 12. FIGURES
# ======================================================================

created_figures = []


# ----------------------------------------------------------------------
# Test RMSE comparison
# ----------------------------------------------------------------------

plt.figure(
    figsize=(
        13,
        7
    )
)


plot_labels = (
    test_results_df[
        "Experiment"
    ]
    + "\n"
    + test_results_df[
        "Model"
    ]
)


plt.bar(
    plot_labels,
    test_results_df[
        "Test_RMSE"
    ]
)


plt.title(
    "Test RMSE Comparison (2022)"
)

plt.xlabel(
    "Experiment and Model"
)

plt.ylabel(
    "RMSE (mm)"
)

plt.xticks(
    rotation=45,
    ha="right"
)

plt.grid(
    axis="y",
    alpha=0.25
)


created_figures.append(
    save_figure(
        "01_test_rmse_comparison.png"
    )
)


# ----------------------------------------------------------------------
# Test R2 comparison
# ----------------------------------------------------------------------

plt.figure(
    figsize=(
        13,
        7
    )
)


plt.bar(
    plot_labels,
    test_results_df[
        "Test_R2"
    ]
)


plt.title(
    "Test R² Comparison (2022)"
)

plt.xlabel(
    "Experiment and Model"
)

plt.ylabel(
    "R²"
)

plt.xticks(
    rotation=45,
    ha="right"
)

plt.grid(
    axis="y",
    alpha=0.25
)


created_figures.append(
    save_figure(
        "02_test_r2_comparison.png"
    )
)


# ----------------------------------------------------------------------
# Observed vs predicted scatter for best model
# ----------------------------------------------------------------------

plt.figure(
    figsize=(
        8,
        8
    )
)


plt.scatter(
    best_prediction_df[
        "Observed_Rainfall_mm"
    ],
    best_prediction_df[
        "Predicted_Rainfall_mm"
    ],
    alpha=0.75
)


minimum_value = min(
    best_prediction_df[
        "Observed_Rainfall_mm"
    ].min(),
    best_prediction_df[
        "Predicted_Rainfall_mm"
    ].min(),
)


maximum_value = max(
    best_prediction_df[
        "Observed_Rainfall_mm"
    ].max(),
    best_prediction_df[
        "Predicted_Rainfall_mm"
    ].max(),
)


plt.plot(
    [
        minimum_value,
        maximum_value
    ],
    [
        minimum_value,
        maximum_value
    ],
    linestyle="--"
)


plt.title(
    "Observed vs Predicted Rainfall\n"
    f"{best_test_row['Experiment']} | "
    f"{best_test_row['Model']}"
)

plt.xlabel(
    "Observed rainfall (mm)"
)

plt.ylabel(
    "Predicted rainfall (mm)"
)

plt.grid(
    alpha=0.25
)


created_figures.append(
    save_figure(
        "03_best_model_observed_vs_predicted.png"
    )
)


# ----------------------------------------------------------------------
# Residual distribution
# ----------------------------------------------------------------------

plt.figure(
    figsize=(
        9,
        6
    )
)


plt.hist(
    best_prediction_df[
        "Residual_mm"
    ],
    bins=20
)


plt.title(
    "Residual Distribution of Best Test Model"
)

plt.xlabel(
    "Residual (Predicted - Observed) mm"
)

plt.ylabel(
    "Frequency"
)

plt.grid(
    alpha=0.25
)


created_figures.append(
    save_figure(
        "04_best_model_residual_distribution.png"
    )
)


# ----------------------------------------------------------------------
# Best model feature importance
# ----------------------------------------------------------------------

best_model_key = (
    f"{safe_name(best_test_row['Experiment'])}"
    f"__{safe_name(best_test_row['Model'])}"
)


best_importance_file = (
    IMPORTANCE_FOLDER
    / f"{best_model_key}__permutation_importance.csv"
)


if best_importance_file.exists():

    best_importance_df = pd.read_csv(
        best_importance_file
    )


    best_importance_plot_df = (
        best_importance_df
        .head(20)
        .sort_values(
            "Importance",
            ascending=True
        )
    )


    plt.figure(
        figsize=(
            10,
            8
        )
    )


    plt.barh(
        best_importance_plot_df[
            "Feature"
        ],
        best_importance_plot_df[
            "Importance"
        ]
    )


    plt.title(
        "Permutation Feature Importance of Best Model"
    )

    plt.xlabel(
        "Decrease in predictive performance"
    )

    plt.ylabel(
        "Feature"
    )

    plt.grid(
        axis="x",
        alpha=0.25
    )


    created_figures.append(
        save_figure(
            "05_best_model_feature_importance.png"
        )
    )


# ======================================================================
# 13. SAVE OUTPUT TABLES
# ======================================================================

CANDIDATE_RESULTS_OUTPUT = (
    RESULT_FOLDER
    / "all_validation_candidates.csv"
)


VALIDATION_COMPARISON_OUTPUT = (
    RESULT_FOLDER
    / "model_comparison_validation.csv"
)


TEST_COMPARISON_OUTPUT = (
    RESULT_FOLDER
    / "model_comparison_test.csv"
)


BEST_MODEL_SUMMARY_OUTPUT = (
    RESULT_FOLDER
    / "best_model_summary.csv"
)


ALL_TEST_PREDICTIONS_OUTPUT = (
    PREDICTION_FOLDER
    / "all_test_predictions.csv"
)


BEST_TEST_PREDICTIONS_OUTPUT = (
    PREDICTION_FOLDER
    / "best_model_test_predictions.csv"
)


BEST_STATION_METRICS_OUTPUT = (
    RESULT_FOLDER
    / "best_model_station_metrics.csv"
)


BEST_MONTHLY_METRICS_OUTPUT = (
    RESULT_FOLDER
    / "best_model_monthly_metrics.csv"
)


EXPERIMENT_INVENTORY_OUTPUT = (
    LOG_FOLDER
    / "model_training_experiment_inventory.csv"
)


OPTIONAL_MODEL_STATUS_OUTPUT = (
    LOG_FOLDER
    / "model_training_optional_model_status.csv"
)


candidate_results_df.to_csv(
    CANDIDATE_RESULTS_OUTPUT,
    index=False
)


validation_best_df.to_csv(
    VALIDATION_COMPARISON_OUTPUT,
    index=False
)


test_results_df.to_csv(
    TEST_COMPARISON_OUTPUT,
    index=False
)


best_model_summary_df.to_csv(
    BEST_MODEL_SUMMARY_OUTPUT,
    index=False
)


all_test_predictions_df.to_csv(
    ALL_TEST_PREDICTIONS_OUTPUT,
    index=False
)


best_prediction_df.to_csv(
    BEST_TEST_PREDICTIONS_OUTPUT,
    index=False
)


station_metrics_df.to_csv(
    BEST_STATION_METRICS_OUTPUT,
    index=False
)


monthly_metrics_df.to_csv(
    BEST_MONTHLY_METRICS_OUTPUT,
    index=False
)


experiment_inventory_df.to_csv(
    EXPERIMENT_INVENTORY_OUTPUT,
    index=False
)


pd.DataFrame([
    {
        "Model": model_name,
        "Available": available,
    }
    for model_name, available
    in MODEL_AVAILABILITY.items()
]).to_csv(
    OPTIONAL_MODEL_STATUS_OUTPUT,
    index=False
)


# ======================================================================
# 14. DISPLAY RESULTS
# ======================================================================

print()
print("=" * 88)
print("BEST VALIDATION CONFIGURATIONS")
print("=" * 88)

display(
    validation_best_df
)


print()
print("=" * 88)
print("TEST MODEL COMPARISON")
print("=" * 88)

display(
    test_results_df
)


print()
print("=" * 88)
print("BEST MODEL SUMMARY")
print("=" * 88)

display(
    best_model_summary_df
)


print()
print("=" * 88)
print("BEST MODEL STATION METRICS")
print("=" * 88)

display(
    station_metrics_df
)


# ======================================================================
# 15. FINAL COMPLETION MESSAGE
# ======================================================================

print()
print("=" * 88)
print("MODEL TRAINING COMPLETED")
print("=" * 88)


print(
    f"Experiments loaded : "
    f"{len(experiment_data)}"
)


print(
    f"Models evaluated   : "
    f"{len(MODEL_PARAM_GRIDS)}"
)


print(
    f"Successful candidates: "
    f"{len(successful_candidate_df)}"
)


print(
    f"Best experiment    : "
    f"{best_test_row['Experiment']}"
)


print(
    f"Best model         : "
    f"{best_test_row['Model']}"
)


print(
    f"Test RMSE          : "
    f"{best_test_row['Test_RMSE']:.4f}"
)


print(
    f"Test MAE           : "
    f"{best_test_row['Test_MAE']:.4f}"
)


print(
    f"Test R²            : "
    f"{best_test_row['Test_R2']:.4f}"
)


print(
    f"Test Bias          : "
    f"{best_test_row['Test_Bias']:.4f}"
)


print(
    f"Test Pearson R     : "
    f"{best_test_row['Test_Pearson_R']:.4f}"
)


print(
    f"Test NSE           : "
    f"{best_test_row['Test_NSE']:.4f}"
)


print(
    f"Test KGE           : "
    f"{best_test_row['Test_KGE']:.4f}"
)


print(
    f"Figures created    : "
    f"{len(created_figures)}"
)


print()
print(
    "Best saved model:"
)

print(
    best_test_row[
        "Saved_Model"
    ]
)


print()
print(
    "Main result files:"
)

print(
    f"1. {VALIDATION_COMPARISON_OUTPUT}"
)

print(
    f"2. {TEST_COMPARISON_OUTPUT}"
)

print(
    f"3. {BEST_MODEL_SUMMARY_OUTPUT}"
)

print(
    f"4. {BEST_TEST_PREDICTIONS_OUTPUT}"
)

print(
    f"5. {BEST_STATION_METRICS_OUTPUT}"
)

print(
    f"6. {BEST_MONTHLY_METRICS_OUTPUT}"
)
```
