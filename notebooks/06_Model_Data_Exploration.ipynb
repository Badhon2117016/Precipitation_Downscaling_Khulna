```python
# ======================================================================
# 06_Model_Data_Exploration.ipynb
# FULL MODEL DATA EXPLORATION AND QUALITY-CONTROL PIPELINE
# ======================================================================
#
# Input:
#   data/processed/model_data/gauge_predictor_dataset_clean.csv
#
# Output:
#   data/processed/model_data/
#       model_dataset_eda_ready.csv
#       model_dataset_without_extreme_outliers.csv
#
# Logs:
#   data/processed/logs/
#       model_eda_overall_summary.csv
#       model_eda_missing_summary.csv
#       model_eda_numeric_summary.csv
#       model_eda_predictor_quality.csv
#       model_eda_outlier_summary.csv
#       model_eda_high_correlation_pairs.csv
#       model_eda_vif.csv
#       model_eda_station_summary.csv
#       model_eda_monthly_summary.csv
#       model_eda_seasonal_summary.csv
#       model_eda_annual_summary.csv
#
# Figures:
#   figures/model_eda/
# ======================================================================


# ======================================================================
# 1. IMPORT LIBRARIES
# ======================================================================

from pathlib import Path
import re
import warnings

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt

from statsmodels.stats.outliers_influence import variance_inflation_factor


warnings.filterwarnings("ignore")


# ======================================================================
# 2. PROJECT CONFIGURATION
# ======================================================================

PROJECT_ROOT = Path(
    r"E:\Geospatial\Precipitation Downscaling"
    r"\Precipitation-Downscaling-Khulna"
)


INPUT_CSV = (
    PROJECT_ROOT
    / "data"
    / "processed"
    / "model_data"
    / "gauge_predictor_dataset_clean.csv"
)


MODEL_DATA_FOLDER = (
    PROJECT_ROOT
    / "data"
    / "processed"
    / "model_data"
)


LOG_FOLDER = (
    PROJECT_ROOT
    / "data"
    / "processed"
    / "logs"
)


FIGURE_FOLDER = (
    PROJECT_ROOT
    / "figures"
    / "model_eda"
)


MODEL_DATA_FOLDER.mkdir(
    parents=True,
    exist_ok=True
)

LOG_FOLDER.mkdir(
    parents=True,
    exist_ok=True
)

FIGURE_FOLDER.mkdir(
    parents=True,
    exist_ok=True
)


# ======================================================================
# 3. ANALYSIS SETTINGS
# ======================================================================

TARGET_COLUMN = "rainfall_mm"


EXPECTED_PREDICTORS = [
    "CCS",
    "CDR",
    "CHIRPS",
    "ERA5",
    "GSMaP_Gauge",
    "GSMaP_MVK",
    "IMERG",
    "LST",
    "NDVI",
    "PDIR",
    "PERSIANN",
    "Distance_Sea_meter",
    "Land_Aspect_Degree_30m",
    "Land_Khulna_SRTM_DEM",
    "Land_Slope_Degree_30m",
]


TEMPORAL_FEATURES = [
    "month_sin",
    "month_cos",
]


COORDINATE_FEATURES = [
    "latitude",
    "longitude",
]


IDENTIFIER_COLUMNS = [
    "station_id",
    "station_name",
    "date",
    "month_id",
    "season",
]


QUALITY_COLUMNS = [
    "raster_x",
    "raster_y",
    "inside_raster_bounds",
    "valid_predictor_count",
    "missing_predictor_count",
    "all_predictors_available",
]


# Correlation threshold
HIGH_CORRELATION_THRESHOLD = 0.85


# VIF threshold
HIGH_VIF_THRESHOLD = 10.0


# Near-constant threshold
NEAR_CONSTANT_DOMINANT_RATIO = 0.95


# IQR multiplier for outlier detection
IQR_MULTIPLIER = 1.5


# Extreme outlier multiplier
EXTREME_IQR_MULTIPLIER = 3.0


# Whether to create a second dataset without extreme target outliers
CREATE_EXTREME_OUTLIER_FILTERED_DATASET = True


# Whether to include gauge-adjusted product in the main analysis
INCLUDE_GSMAP_GAUGE = True


# ======================================================================
# 4. HELPER FUNCTIONS
# ======================================================================

def normalize_column_name(column_name):
    """
    Convert a column name into a consistent snake_case-like format
    while preserving capitalization where useful.
    """

    column_name = str(column_name).strip()

    column_name = re.sub(
        r"\s+",
        "_",
        column_name
    )

    column_name = re.sub(
        r"[^A-Za-z0-9_]+",
        "",
        column_name
    )

    return column_name


def safe_numeric_conversion(dataframe, columns):
    """
    Convert selected columns to numeric safely.
    """

    dataframe = dataframe.copy()

    for column in columns:

        if column in dataframe.columns:

            dataframe[column] = pd.to_numeric(
                dataframe[column],
                errors="coerce"
            )

    return dataframe


def assign_season(month):
    """
    Bangladesh seasonal grouping.
    """

    if month in [12, 1, 2]:
        return "Winter"

    if month in [3, 4, 5]:
        return "Pre_Monsoon"

    if month in [6, 7, 8, 9]:
        return "Monsoon"

    return "Post_Monsoon"


def calculate_iqr_bounds(series, multiplier=1.5):
    """
    Calculate lower and upper IQR bounds.
    """

    clean_series = series.dropna()

    if clean_series.empty:

        return np.nan, np.nan, np.nan, np.nan

    q1 = clean_series.quantile(0.25)
    q3 = clean_series.quantile(0.75)

    iqr = q3 - q1

    lower_bound = q1 - multiplier * iqr
    upper_bound = q3 + multiplier * iqr

    return q1, q3, lower_bound, upper_bound


def count_iqr_outliers(series, multiplier=1.5):
    """
    Count IQR-based outliers.
    """

    _, _, lower_bound, upper_bound = calculate_iqr_bounds(
        series,
        multiplier
    )

    if pd.isna(lower_bound) or pd.isna(upper_bound):

        return 0

    return int(
        (
            (series < lower_bound)
            |
            (series > upper_bound)
        ).sum()
    )


def detect_constant_and_near_constant(series):
    """
    Detect constant and near-constant variables.
    """

    clean_series = series.dropna()

    if clean_series.empty:

        return {
            "Unique_Values": 0,
            "Constant": True,
            "Dominant_Ratio": np.nan,
            "Near_Constant": True,
        }

    unique_values = clean_series.nunique()

    value_counts = clean_series.value_counts(
        normalize=True,
        dropna=True
    )

    dominant_ratio = (
        float(value_counts.iloc[0])
        if not value_counts.empty
        else np.nan
    )

    constant = unique_values <= 1

    near_constant = (
        constant
        or dominant_ratio >= NEAR_CONSTANT_DOMINANT_RATIO
    )

    return {
        "Unique_Values": int(unique_values),
        "Constant": bool(constant),
        "Dominant_Ratio": dominant_ratio,
        "Near_Constant": bool(near_constant),
    }


def calculate_high_correlation_pairs(
    correlation_matrix,
    threshold=0.85
):
    """
    Return unique high-correlation feature pairs.
    """

    records = []

    columns = list(
        correlation_matrix.columns
    )

    for i in range(len(columns)):

        for j in range(i + 1, len(columns)):

            feature_1 = columns[i]
            feature_2 = columns[j]

            correlation = correlation_matrix.loc[
                feature_1,
                feature_2
            ]

            if pd.isna(correlation):
                continue

            if abs(correlation) >= threshold:

                records.append({
                    "Feature_1": feature_1,
                    "Feature_2": feature_2,
                    "Correlation": float(correlation),
                    "Absolute_Correlation": float(
                        abs(correlation)
                    ),
                })

    if not records:

        return pd.DataFrame(
            columns=[
                "Feature_1",
                "Feature_2",
                "Correlation",
                "Absolute_Correlation",
            ]
        )

    output_df = pd.DataFrame(
        records
    )

    return output_df.sort_values(
        "Absolute_Correlation",
        ascending=False
    ).reset_index(
        drop=True
    )


def calculate_vif(dataframe, columns):
    """
    Calculate VIF for numeric predictors.

    Constant columns and invalid columns are excluded.
    """

    available_columns = [
        column
        for column in columns
        if column in dataframe.columns
    ]

    vif_data = dataframe[
        available_columns
    ].copy()

    vif_data = vif_data.replace(
        [np.inf, -np.inf],
        np.nan
    )

    vif_data = vif_data.dropna()

    usable_columns = []

    for column in vif_data.columns:

        if vif_data[column].nunique() > 1:
            usable_columns.append(column)

    vif_data = vif_data[
        usable_columns
    ]


    if vif_data.empty or len(usable_columns) < 2:

        return pd.DataFrame(
            columns=[
                "Feature",
                "VIF",
                "High_VIF",
            ]
        )


    vif_records = []

    values = vif_data.astype(float).values


    for index, column in enumerate(
        usable_columns
    ):

        try:

            vif_value = variance_inflation_factor(
                values,
                index
            )

        except Exception:

            vif_value = np.nan


        vif_records.append({
            "Feature": column,
            "VIF": float(vif_value)
            if np.isfinite(vif_value)
            else np.inf,
            "High_VIF": bool(
                vif_value >= HIGH_VIF_THRESHOLD
            )
            if np.isfinite(vif_value)
            else True,
        })


    return (
        pd.DataFrame(vif_records)
        .sort_values(
            "VIF",
            ascending=False
        )
        .reset_index(
            drop=True
        )
    )


def save_figure(filename):
    """
    Save current matplotlib figure.
    """

    output_path = (
        FIGURE_FOLDER
        / filename
    )

    plt.tight_layout()

    plt.savefig(
        output_path,
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()

    plt.close()

    return output_path


def create_histogram(
    dataframe,
    column,
    title,
    xlabel,
    filename,
    bins=30
):
    """
    Create and save a histogram.
    """

    clean_data = dataframe[
        column
    ].dropna()

    if clean_data.empty:
        return None

    plt.figure(
        figsize=(9, 6)
    )

    plt.hist(
        clean_data,
        bins=bins
    )

    plt.title(
        title
    )

    plt.xlabel(
        xlabel
    )

    plt.ylabel(
        "Frequency"
    )

    plt.grid(
        alpha=0.25
    )

    return save_figure(
        filename
    )


def create_boxplot(
    dataframe,
    column,
    title,
    ylabel,
    filename
):
    """
    Create and save a boxplot.
    """

    clean_data = dataframe[
        column
    ].dropna()

    if clean_data.empty:
        return None

    plt.figure(
        figsize=(8, 6)
    )

    plt.boxplot(
        clean_data,
        vert=True
    )

    plt.title(
        title
    )

    plt.ylabel(
        ylabel
    )

    plt.grid(
        axis="y",
        alpha=0.25
    )

    return save_figure(
        filename
    )


# ======================================================================
# 5. INITIAL VALIDATION
# ======================================================================

print("=" * 82)
print("MODEL DATA EXPLORATION AND QUALITY CONTROL")
print("=" * 82)

print(f"Project root : {PROJECT_ROOT}")
print(f"Input CSV    : {INPUT_CSV}")
print(f"Log folder   : {LOG_FOLDER}")
print(f"Figure folder: {FIGURE_FOLDER}")
print()


if not PROJECT_ROOT.exists():

    raise FileNotFoundError(
        f"Project root পাওয়া যায়নি:\n"
        f"{PROJECT_ROOT}"
    )


if not INPUT_CSV.exists():

    raise FileNotFoundError(
        f"Model dataset পাওয়া যায়নি:\n"
        f"{INPUT_CSV}"
    )


# ======================================================================
# 6. LOAD DATASET
# ======================================================================

print("=" * 82)
print("LOADING MODEL DATASET")
print("=" * 82)


df = pd.read_csv(
    INPUT_CSV
)


df.columns = [
    normalize_column_name(column)
    for column in df.columns
]


print(
    f"Rows    : {len(df)}"
)

print(
    f"Columns : {len(df.columns)}"
)

print()
print("Columns:")

for number, column in enumerate(
    df.columns,
    start=1
):

    print(
        f"{number:02d}. {column}"
    )


# ======================================================================
# 7. STANDARDIZE DATA TYPES
# ======================================================================

if "date" in df.columns:

    df["date"] = pd.to_datetime(
        df["date"],
        errors="coerce"
    )


numeric_candidate_columns = (
    EXPECTED_PREDICTORS
    + TEMPORAL_FEATURES
    + COORDINATE_FEATURES
    + [
        TARGET_COLUMN,
        "year",
        "month",
        "raster_x",
        "raster_y",
        "valid_predictor_count",
        "missing_predictor_count",
    ]
)


df = safe_numeric_conversion(
    df,
    numeric_candidate_columns
)


if "season" not in df.columns:

    df["season"] = df[
        "month"
    ].apply(
        assign_season
    )


if "month_id" not in df.columns:

    df["month_id"] = (
        df["year"]
        .astype("Int64")
        .astype(str)
        + "_"
        + df["month"]
        .astype("Int64")
        .astype(str)
        .str.zfill(2)
    )


# ======================================================================
# 8. REQUIRED COLUMN VALIDATION
# ======================================================================

required_columns = [
    "station_id",
    "station_name",
    "latitude",
    "longitude",
    "year",
    "month",
    TARGET_COLUMN,
]


missing_required_columns = [
    column
    for column in required_columns
    if column not in df.columns
]


if missing_required_columns:

    raise ValueError(
        "Required column missing:\n"
        + ", ".join(
            missing_required_columns
        )
    )


available_predictors = [
    column
    for column in EXPECTED_PREDICTORS
    if column in df.columns
]


missing_expected_predictors = [
    column
    for column in EXPECTED_PREDICTORS
    if column not in df.columns
]


if not INCLUDE_GSMAP_GAUGE:

    available_predictors = [
        column
        for column in available_predictors
        if column != "GSMaP_Gauge"
    ]


print()
print("=" * 82)
print("PREDICTOR INVENTORY")
print("=" * 82)

print(
    f"Expected predictors : "
    f"{len(EXPECTED_PREDICTORS)}"
)

print(
    f"Available predictors: "
    f"{len(available_predictors)}"
)


if missing_expected_predictors:

    print()
    print("Missing expected predictors:")

    for column in missing_expected_predictors:
        print(f"  {column}")


# ======================================================================
# 9. BASIC DATA QUALITY CHECKS
# ======================================================================

print()
print("=" * 82)
print("BASIC DATA QUALITY")
print("=" * 82)


duplicate_count = int(
    df.duplicated(
        subset=[
            "station_id",
            "year",
            "month"
        ]
    ).sum()
)


missing_target_count = int(
    df[
        TARGET_COLUMN
    ].isna().sum()
)


negative_target_count = int(
    (
        df[
            TARGET_COLUMN
        ]
        < 0
    ).sum()
)


invalid_month_count = int(
    (
        ~df[
            "month"
        ].between(
            1,
            12
        )
    ).sum()
)


invalid_latitude_count = int(
    (
        ~df[
            "latitude"
        ].between(
            -90,
            90
        )
    ).sum()
)


invalid_longitude_count = int(
    (
        ~df[
            "longitude"
        ].between(
            -180,
            180
        )
    ).sum()
)


print(
    f"Duplicate station-month rows : "
    f"{duplicate_count}"
)

print(
    f"Missing target rainfall      : "
    f"{missing_target_count}"
)

print(
    f"Negative target rainfall     : "
    f"{negative_target_count}"
)

print(
    f"Invalid month values         : "
    f"{invalid_month_count}"
)

print(
    f"Invalid latitude values      : "
    f"{invalid_latitude_count}"
)

print(
    f"Invalid longitude values     : "
    f"{invalid_longitude_count}"
)

print(
    f"Unique stations              : "
    f"{df['station_id'].nunique()}"
)

print(
    f"Year range                   : "
    f"{df['year'].min()} to "
    f"{df['year'].max()}"
)


# ======================================================================
# 10. MISSING VALUE ANALYSIS
# ======================================================================

missing_summary_df = pd.DataFrame({
    "Column": df.columns,
    "Missing_Values": [
        int(
            df[column]
            .isna()
            .sum()
        )
        for column in df.columns
    ],
})


missing_summary_df[
    "Missing_Percentage"
] = (
    missing_summary_df[
        "Missing_Values"
    ]
    /
    len(df)
    * 100
)


missing_summary_df[
    "Data_Type"
] = [
    str(
        df[column].dtype
    )
    for column in df.columns
]


missing_summary_df = (
    missing_summary_df
    .sort_values(
        [
            "Missing_Percentage",
            "Column"
        ],
        ascending=[
            False,
            True
        ]
    )
    .reset_index(
        drop=True
    )
)


print()
print("=" * 82)
print("MISSING VALUE SUMMARY")
print("=" * 82)

display(
    missing_summary_df
)


# ======================================================================
# 11. NUMERIC SUMMARY
# ======================================================================

numeric_columns = df.select_dtypes(
    include=[
        np.number
    ]
).columns.tolist()


numeric_summary_df = (
    df[
        numeric_columns
    ]
    .describe(
        percentiles=[
            0.01,
            0.05,
            0.25,
            0.50,
            0.75,
            0.95,
            0.99,
        ]
    )
    .T
    .reset_index()
    .rename(
        columns={
            "index": "Variable"
        }
    )
)


numeric_summary_df[
    "Missing"
] = [
    int(
        df[column]
        .isna()
        .sum()
    )
    for column in numeric_summary_df[
        "Variable"
    ]
]


numeric_summary_df[
    "Unique"
] = [
    int(
        df[column]
        .nunique(
            dropna=True
        )
    )
    for column in numeric_summary_df[
        "Variable"
    ]
]


print()
print("=" * 82)
print("NUMERIC SUMMARY")
print("=" * 82)

display(
    numeric_summary_df
)


# ======================================================================
# 12. PREDICTOR QUALITY ANALYSIS
# ======================================================================

predictor_quality_records = []


for predictor in available_predictors:

    series = df[
        predictor
    ]


    constant_info = detect_constant_and_near_constant(
        series
    )


    negative_count = int(
        (
            series < 0
        ).sum()
    )


    zero_count = int(
        (
            series == 0
        ).sum()
    )


    missing_count = int(
        series.isna().sum()
    )


    finite_count = int(
        np.isfinite(
            series.dropna()
        ).sum()
    )


    q1, q3, lower_bound, upper_bound = calculate_iqr_bounds(
        series,
        IQR_MULTIPLIER
    )


    outlier_count = count_iqr_outliers(
        series,
        IQR_MULTIPLIER
    )


    extreme_outlier_count = count_iqr_outliers(
        series,
        EXTREME_IQR_MULTIPLIER
    )


    predictor_quality_records.append({
        "Predictor": predictor,
        "Minimum": series.min(),
        "Maximum": series.max(),
        "Mean": series.mean(),
        "Median": series.median(),
        "Standard_Deviation": series.std(),
        "Missing_Values": missing_count,
        "Finite_Values": finite_count,
        "Negative_Values": negative_count,
        "Zero_Values": zero_count,
        "Unique_Values": constant_info[
            "Unique_Values"
        ],
        "Constant": constant_info[
            "Constant"
        ],
        "Dominant_Ratio": constant_info[
            "Dominant_Ratio"
        ],
        "Near_Constant": constant_info[
            "Near_Constant"
        ],
        "Q1": q1,
        "Q3": q3,
        "IQR_Lower_Bound": lower_bound,
        "IQR_Upper_Bound": upper_bound,
        "IQR_Outliers": outlier_count,
        "Extreme_IQR_Outliers": (
            extreme_outlier_count
        ),
    })


predictor_quality_df = pd.DataFrame(
    predictor_quality_records
)


print()
print("=" * 82)
print("PREDICTOR QUALITY SUMMARY")
print("=" * 82)

display(
    predictor_quality_df
)


# ======================================================================
# 13. TARGET RAINFALL EXPLORATION
# ======================================================================

target_series = df[
    TARGET_COLUMN
]


target_q1, target_q3, target_lower, target_upper = (
    calculate_iqr_bounds(
        target_series,
        IQR_MULTIPLIER
    )
)


(
    target_extreme_q1,
    target_extreme_q3,
    target_extreme_lower,
    target_extreme_upper
) = calculate_iqr_bounds(
    target_series,
    EXTREME_IQR_MULTIPLIER
)


target_outlier_count = count_iqr_outliers(
    target_series,
    IQR_MULTIPLIER
)


target_extreme_outlier_count = count_iqr_outliers(
    target_series,
    EXTREME_IQR_MULTIPLIER
)


dry_month_count = int(
    (
        target_series == 0
    ).sum()
)


target_summary_df = pd.DataFrame([
    {
        "Metric": "Count",
        "Value": int(
            target_series.count()
        ),
    },
    {
        "Metric": "Missing",
        "Value": int(
            target_series.isna().sum()
        ),
    },
    {
        "Metric": "Minimum",
        "Value": target_series.min(),
    },
    {
        "Metric": "Maximum",
        "Value": target_series.max(),
    },
    {
        "Metric": "Mean",
        "Value": target_series.mean(),
    },
    {
        "Metric": "Median",
        "Value": target_series.median(),
    },
    {
        "Metric": "Standard deviation",
        "Value": target_series.std(),
    },
    {
        "Metric": "Dry months",
        "Value": dry_month_count,
    },
    {
        "Metric": "IQR outliers",
        "Value": target_outlier_count,
    },
    {
        "Metric": "Extreme IQR outliers",
        "Value": target_extreme_outlier_count,
    },
    {
        "Metric": "IQR lower bound",
        "Value": target_lower,
    },
    {
        "Metric": "IQR upper bound",
        "Value": target_upper,
    },
    {
        "Metric": "Extreme lower bound",
        "Value": target_extreme_lower,
    },
    {
        "Metric": "Extreme upper bound",
        "Value": target_extreme_upper,
    },
])


print()
print("=" * 82)
print("TARGET RAINFALL SUMMARY")
print("=" * 82)

display(
    target_summary_df
)


# ======================================================================
# 14. OUTLIER SUMMARY
# ======================================================================

outlier_columns = (
    [TARGET_COLUMN]
    + available_predictors
)


outlier_summary_records = []


for column in outlier_columns:

    series = df[
        column
    ]


    q1, q3, lower_bound, upper_bound = calculate_iqr_bounds(
        series,
        IQR_MULTIPLIER
    )


    (
        extreme_q1,
        extreme_q3,
        extreme_lower,
        extreme_upper
    ) = calculate_iqr_bounds(
        series,
        EXTREME_IQR_MULTIPLIER
    )


    regular_outliers = count_iqr_outliers(
        series,
        IQR_MULTIPLIER
    )


    extreme_outliers = count_iqr_outliers(
        series,
        EXTREME_IQR_MULTIPLIER
    )


    outlier_summary_records.append({
        "Variable": column,
        "Q1": q1,
        "Q3": q3,
        "IQR_Lower_Bound": lower_bound,
        "IQR_Upper_Bound": upper_bound,
        "IQR_Outlier_Count": regular_outliers,
        "Extreme_Lower_Bound": extreme_lower,
        "Extreme_Upper_Bound": extreme_upper,
        "Extreme_Outlier_Count": extreme_outliers,
    })


outlier_summary_df = pd.DataFrame(
    outlier_summary_records
)


print()
print("=" * 82)
print("OUTLIER SUMMARY")
print("=" * 82)

display(
    outlier_summary_df
)


# ======================================================================
# 15. STATION-WISE SUMMARY
# ======================================================================

station_summary_df = (
    df
    .groupby(
        [
            "station_id",
            "station_name"
        ],
        as_index=False
    )
    .agg(
        records=(
            TARGET_COLUMN,
            "size"
        ),
        valid_rainfall=(
            TARGET_COLUMN,
            "count"
        ),
        missing_rainfall=(
            TARGET_COLUMN,
            lambda series: int(
                series.isna().sum()
            )
        ),
        mean_rainfall_mm=(
            TARGET_COLUMN,
            "mean"
        ),
        median_rainfall_mm=(
            TARGET_COLUMN,
            "median"
        ),
        min_rainfall_mm=(
            TARGET_COLUMN,
            "min"
        ),
        max_rainfall_mm=(
            TARGET_COLUMN,
            "max"
        ),
        std_rainfall_mm=(
            TARGET_COLUMN,
            "std"
        ),
        latitude=(
            "latitude",
            "first"
        ),
        longitude=(
            "longitude",
            "first"
        ),
    )
)


station_summary_df[
    "completeness_percent"
] = (
    station_summary_df[
        "valid_rainfall"
    ]
    /
    station_summary_df[
        "records"
    ]
    * 100
)


print()
print("=" * 82)
print("STATION SUMMARY")
print("=" * 82)

display(
    station_summary_df
)


# ======================================================================
# 16. MONTHLY CLIMATOLOGY
# ======================================================================

monthly_summary_df = (
    df
    .groupby(
        "month",
        as_index=False
    )
    .agg(
        records=(
            TARGET_COLUMN,
            "size"
        ),
        mean_rainfall_mm=(
            TARGET_COLUMN,
            "mean"
        ),
        median_rainfall_mm=(
            TARGET_COLUMN,
            "median"
        ),
        min_rainfall_mm=(
            TARGET_COLUMN,
            "min"
        ),
        max_rainfall_mm=(
            TARGET_COLUMN,
            "max"
        ),
        std_rainfall_mm=(
            TARGET_COLUMN,
            "std"
        ),
    )
    .sort_values(
        "month"
    )
    .reset_index(
        drop=True
    )
)


print()
print("=" * 82)
print("MONTHLY RAINFALL SUMMARY")
print("=" * 82)

display(
    monthly_summary_df
)


# ======================================================================
# 17. SEASONAL SUMMARY
# ======================================================================

season_order = [
    "Winter",
    "Pre_Monsoon",
    "Monsoon",
    "Post_Monsoon",
]


seasonal_summary_df = (
    df
    .groupby(
        "season",
        as_index=False
    )
    .agg(
        records=(
            TARGET_COLUMN,
            "size"
        ),
        mean_rainfall_mm=(
            TARGET_COLUMN,
            "mean"
        ),
        median_rainfall_mm=(
            TARGET_COLUMN,
            "median"
        ),
        min_rainfall_mm=(
            TARGET_COLUMN,
            "min"
        ),
        max_rainfall_mm=(
            TARGET_COLUMN,
            "max"
        ),
        std_rainfall_mm=(
            TARGET_COLUMN,
            "std"
        ),
    )
)


seasonal_summary_df[
    "season"
] = pd.Categorical(
    seasonal_summary_df[
        "season"
    ],
    categories=season_order,
    ordered=True
)


seasonal_summary_df = (
    seasonal_summary_df
    .sort_values(
        "season"
    )
    .reset_index(
        drop=True
    )
)


print()
print("=" * 82)
print("SEASONAL RAINFALL SUMMARY")
print("=" * 82)

display(
    seasonal_summary_df
)


# ======================================================================
# 18. ANNUAL SUMMARY
# ======================================================================

annual_station_df = (
    df
    .groupby(
        [
            "year",
            "station_id"
        ],
        as_index=False
    )
    .agg(
        annual_total_rainfall_mm=(
            TARGET_COLUMN,
            "sum"
        ),
        annual_mean_monthly_rainfall_mm=(
            TARGET_COLUMN,
            "mean"
        ),
        valid_months=(
            TARGET_COLUMN,
            "count"
        ),
    )
)


annual_summary_df = (
    annual_station_df
    .groupby(
        "year",
        as_index=False
    )
    .agg(
        stations=(
            "station_id",
            "nunique"
        ),
        mean_annual_total_rainfall_mm=(
            "annual_total_rainfall_mm",
            "mean"
        ),
        min_annual_total_rainfall_mm=(
            "annual_total_rainfall_mm",
            "min"
        ),
        max_annual_total_rainfall_mm=(
            "annual_total_rainfall_mm",
            "max"
        ),
        mean_monthly_rainfall_mm=(
            "annual_mean_monthly_rainfall_mm",
            "mean"
        ),
    )
)


print()
print("=" * 82)
print("ANNUAL RAINFALL SUMMARY")
print("=" * 82)

display(
    annual_summary_df
)


# ======================================================================
# 19. CORRELATION ANALYSIS
# ======================================================================

correlation_features = (
    available_predictors
    + TEMPORAL_FEATURES
    + COORDINATE_FEATURES
    + [TARGET_COLUMN]
)


correlation_features = [
    column
    for column in correlation_features
    if column in df.columns
]


pearson_correlation_df = (
    df[
        correlation_features
    ]
    .corr(
        method="pearson"
    )
)


spearman_correlation_df = (
    df[
        correlation_features
    ]
    .corr(
        method="spearman"
    )
)


target_pearson_df = (
    pearson_correlation_df[
        [TARGET_COLUMN]
    ]
    .drop(
        index=TARGET_COLUMN
    )
    .rename(
        columns={
            TARGET_COLUMN: "Pearson_Correlation"
        }
    )
)


target_spearman_df = (
    spearman_correlation_df[
        [TARGET_COLUMN]
    ]
    .drop(
        index=TARGET_COLUMN
    )
    .rename(
        columns={
            TARGET_COLUMN: "Spearman_Correlation"
        }
    )
)


target_correlation_df = (
    target_pearson_df
    .join(
        target_spearman_df
    )
    .reset_index()
    .rename(
        columns={
            "index": "Feature"
        }
    )
)


target_correlation_df[
    "Absolute_Pearson"
] = (
    target_correlation_df[
        "Pearson_Correlation"
    ]
    .abs()
)


target_correlation_df = (
    target_correlation_df
    .sort_values(
        "Absolute_Pearson",
        ascending=False
    )
    .reset_index(
        drop=True
    )
)


predictor_correlation_matrix = (
    df[
        available_predictors
    ]
    .corr(
        method="pearson"
    )
)


high_correlation_pairs_df = (
    calculate_high_correlation_pairs(
        predictor_correlation_matrix,
        HIGH_CORRELATION_THRESHOLD
    )
)


print()
print("=" * 82)
print("CORRELATION WITH TARGET")
print("=" * 82)

display(
    target_correlation_df
)


print()
print("=" * 82)
print(
    f"HIGHLY CORRELATED PREDICTOR PAIRS "
    f"(|r| >= {HIGH_CORRELATION_THRESHOLD})"
)
print("=" * 82)

display(
    high_correlation_pairs_df
)


# ======================================================================
# 20. VIF ANALYSIS
# ======================================================================

vif_features = (
    available_predictors
    + TEMPORAL_FEATURES
    + COORDINATE_FEATURES
)


vif_df = calculate_vif(
    df,
    vif_features
)


print()
print("=" * 82)
print("VARIANCE INFLATION FACTOR")
print("=" * 82)

display(
    vif_df
)


# ======================================================================
# 21. CREATE EDA-READY DATASET
# ======================================================================

eda_df = df.copy()


# Replace infinities
eda_df = eda_df.replace(
    [np.inf, -np.inf],
    np.nan
)


# Remove duplicate station-month records if any
eda_df = eda_df.drop_duplicates(
    subset=[
        "station_id",
        "year",
        "month"
    ],
    keep="first"
).copy()


# Remove missing target rows
eda_df = eda_df.dropna(
    subset=[
        TARGET_COLUMN
    ]
).copy()


# Remove negative target rows
eda_df = eda_df[
    eda_df[
        TARGET_COLUMN
    ]
    >= 0
].copy()


# Ensure valid months
eda_df = eda_df[
    eda_df[
        "month"
    ].between(
        1,
        12
    )
].copy()


# Create temporal features if absent
if "month_sin" not in eda_df.columns:

    eda_df[
        "month_sin"
    ] = np.sin(
        2
        * np.pi
        * eda_df[
            "month"
        ]
        / 12
    )


if "month_cos" not in eda_df.columns:

    eda_df[
        "month_cos"
    ] = np.cos(
        2
        * np.pi
        * eda_df[
            "month"
        ]
        / 12
    )


eda_df = eda_df.sort_values(
    [
        "year",
        "month",
        "station_id"
    ]
).reset_index(
    drop=True
)


# ======================================================================
# 22. EXTREME OUTLIER-FILTERED DATASET
# ======================================================================

if CREATE_EXTREME_OUTLIER_FILTERED_DATASET:

    _, _, extreme_lower_bound, extreme_upper_bound = (
        calculate_iqr_bounds(
            eda_df[
                TARGET_COLUMN
            ],
            EXTREME_IQR_MULTIPLIER
        )
    )


    outlier_filtered_df = eda_df[
        (
            eda_df[
                TARGET_COLUMN
            ]
            >= extreme_lower_bound
        )
        &
        (
            eda_df[
                TARGET_COLUMN
            ]
            <= extreme_upper_bound
        )
    ].copy()


    outlier_filtered_df = (
        outlier_filtered_df
        .reset_index(
            drop=True
        )
    )

else:

    outlier_filtered_df = eda_df.copy()


# ======================================================================
# 23. OVERALL SUMMARY
# ======================================================================

overall_summary_df = pd.DataFrame([
    {
        "Metric": "Input rows",
        "Value": len(df),
    },
    {
        "Metric": "Input columns",
        "Value": len(df.columns),
    },
    {
        "Metric": "Unique stations",
        "Value": df[
            "station_id"
        ].nunique(),
    },
    {
        "Metric": "Start year",
        "Value": df[
            "year"
        ].min(),
    },
    {
        "Metric": "End year",
        "Value": df[
            "year"
        ].max(),
    },
    {
        "Metric": "Available predictors",
        "Value": len(
            available_predictors
        ),
    },
    {
        "Metric": "Duplicate station-month rows",
        "Value": duplicate_count,
    },
    {
        "Metric": "Missing target rows",
        "Value": missing_target_count,
    },
    {
        "Metric": "Negative target rows",
        "Value": negative_target_count,
    },
    {
        "Metric": "EDA-ready rows",
        "Value": len(
            eda_df
        ),
    },
    {
        "Metric": "Extreme-outlier-filtered rows",
        "Value": len(
            outlier_filtered_df
        ),
    },
    {
        "Metric": "Target IQR outliers",
        "Value": target_outlier_count,
    },
    {
        "Metric": "Target extreme outliers",
        "Value": target_extreme_outlier_count,
    },
    {
        "Metric": "High-correlation pairs",
        "Value": len(
            high_correlation_pairs_df
        ),
    },
    {
        "Metric": "High-VIF features",
        "Value": int(
            vif_df[
                "High_VIF"
            ].sum()
        )
        if not vif_df.empty
        else 0,
    },
])


print()
print("=" * 82)
print("OVERALL EDA SUMMARY")
print("=" * 82)

display(
    overall_summary_df
)


# ======================================================================
# 24. FIGURES
# ======================================================================

print()
print("=" * 82)
print("CREATING FIGURES")
print("=" * 82)


created_figures = []


# ----------------------------------------------------------------------
# Target histogram
# ----------------------------------------------------------------------

figure_path = create_histogram(
    dataframe=df,
    column=TARGET_COLUMN,
    title="Distribution of Observed Monthly Rainfall",
    xlabel="Observed rainfall (mm)",
    filename="01_rainfall_distribution.png",
    bins=30
)

if figure_path:
    created_figures.append(figure_path)


# ----------------------------------------------------------------------
# Target boxplot
# ----------------------------------------------------------------------

figure_path = create_boxplot(
    dataframe=df,
    column=TARGET_COLUMN,
    title="Observed Monthly Rainfall Boxplot",
    ylabel="Observed rainfall (mm)",
    filename="02_rainfall_boxplot.png"
)

if figure_path:
    created_figures.append(figure_path)


# ----------------------------------------------------------------------
# Mean monthly rainfall plot
# ----------------------------------------------------------------------

plt.figure(
    figsize=(10, 6)
)

plt.plot(
    monthly_summary_df[
        "month"
    ],
    monthly_summary_df[
        "mean_rainfall_mm"
    ],
    marker="o"
)

plt.title(
    "Mean Monthly Rainfall Climatology"
)

plt.xlabel(
    "Month"
)

plt.ylabel(
    "Mean rainfall (mm)"
)

plt.xticks(
    range(1, 13)
)

plt.grid(
    alpha=0.25
)

created_figures.append(
    save_figure(
        "03_monthly_rainfall_climatology.png"
    )
)


# ----------------------------------------------------------------------
# Seasonal mean rainfall
# ----------------------------------------------------------------------

plt.figure(
    figsize=(9, 6)
)

plt.bar(
    seasonal_summary_df[
        "season"
    ].astype(str),
    seasonal_summary_df[
        "mean_rainfall_mm"
    ]
)

plt.title(
    "Mean Rainfall by Season"
)

plt.xlabel(
    "Season"
)

plt.ylabel(
    "Mean rainfall (mm)"
)

plt.xticks(
    rotation=20
)

plt.grid(
    axis="y",
    alpha=0.25
)

created_figures.append(
    save_figure(
        "04_seasonal_mean_rainfall.png"
    )
)


# ----------------------------------------------------------------------
# Annual rainfall variation
# ----------------------------------------------------------------------

plt.figure(
    figsize=(10, 6)
)

plt.plot(
    annual_summary_df[
        "year"
    ],
    annual_summary_df[
        "mean_annual_total_rainfall_mm"
    ],
    marker="o"
)

plt.title(
    "Mean Annual Total Rainfall Across Stations"
)

plt.xlabel(
    "Year"
)

plt.ylabel(
    "Mean annual rainfall total (mm)"
)

plt.grid(
    alpha=0.25
)

created_figures.append(
    save_figure(
        "05_annual_rainfall_variation.png"
    )
)


# ----------------------------------------------------------------------
# Station-wise mean rainfall
# ----------------------------------------------------------------------

plt.figure(
    figsize=(10, 6)
)

plt.bar(
    station_summary_df[
        "station_name"
    ],
    station_summary_df[
        "mean_rainfall_mm"
    ]
)

plt.title(
    "Mean Monthly Rainfall by Station"
)

plt.xlabel(
    "Station"
)

plt.ylabel(
    "Mean rainfall (mm)"
)

plt.xticks(
    rotation=45,
    ha="right"
)

plt.grid(
    axis="y",
    alpha=0.25
)

created_figures.append(
    save_figure(
        "06_station_mean_rainfall.png"
    )
)


# ----------------------------------------------------------------------
# Correlation with target
# ----------------------------------------------------------------------

plot_correlation_df = (
    target_correlation_df
    .sort_values(
        "Pearson_Correlation"
    )
)


plt.figure(
    figsize=(10, 8)
)

plt.barh(
    plot_correlation_df[
        "Feature"
    ],
    plot_correlation_df[
        "Pearson_Correlation"
    ]
)

plt.title(
    "Pearson Correlation of Features with Observed Rainfall"
)

plt.xlabel(
    "Pearson correlation"
)

plt.ylabel(
    "Feature"
)

plt.grid(
    axis="x",
    alpha=0.25
)

created_figures.append(
    save_figure(
        "07_feature_target_correlation.png"
    )
)


# ----------------------------------------------------------------------
# Correlation matrix
# ----------------------------------------------------------------------

plt.figure(
    figsize=(15, 12)
)

correlation_image = plt.imshow(
    pearson_correlation_df,
    aspect="auto"
)

plt.colorbar(
    correlation_image,
    label="Pearson correlation"
)

plt.xticks(
    ticks=np.arange(
        len(
            pearson_correlation_df.columns
        )
    ),
    labels=pearson_correlation_df.columns,
    rotation=90
)

plt.yticks(
    ticks=np.arange(
        len(
            pearson_correlation_df.index
        )
    ),
    labels=pearson_correlation_df.index
)

plt.title(
    "Pearson Correlation Matrix"
)

created_figures.append(
    save_figure(
        "08_pearson_correlation_matrix.png"
    )
)


# ----------------------------------------------------------------------
# Selected predictor histograms
# ----------------------------------------------------------------------

for index, predictor in enumerate(
    available_predictors,
    start=1
):

    figure_path = create_histogram(
        dataframe=df,
        column=predictor,
        title=f"Distribution of {predictor}",
        xlabel=predictor,
        filename=(
            f"predictor_{index:02d}_"
            f"{predictor}_distribution.png"
        ),
        bins=25
    )

    if figure_path:
        created_figures.append(
            figure_path
        )


print(
    f"Total figures created: "
    f"{len(created_figures)}"
)


# ======================================================================
# 25. OUTPUT PATHS
# ======================================================================

EDA_READY_OUTPUT = (
    MODEL_DATA_FOLDER
    / "model_dataset_eda_ready.csv"
)


OUTLIER_FILTERED_OUTPUT = (
    MODEL_DATA_FOLDER
    / "model_dataset_without_extreme_outliers.csv"
)


OVERALL_SUMMARY_OUTPUT = (
    LOG_FOLDER
    / "model_eda_overall_summary.csv"
)


MISSING_SUMMARY_OUTPUT = (
    LOG_FOLDER
    / "model_eda_missing_summary.csv"
)


NUMERIC_SUMMARY_OUTPUT = (
    LOG_FOLDER
    / "model_eda_numeric_summary.csv"
)


PREDICTOR_QUALITY_OUTPUT = (
    LOG_FOLDER
    / "model_eda_predictor_quality.csv"
)


TARGET_SUMMARY_OUTPUT = (
    LOG_FOLDER
    / "model_eda_target_summary.csv"
)


OUTLIER_SUMMARY_OUTPUT = (
    LOG_FOLDER
    / "model_eda_outlier_summary.csv"
)


PEARSON_CORRELATION_OUTPUT = (
    LOG_FOLDER
    / "model_eda_pearson_correlation.csv"
)


SPEARMAN_CORRELATION_OUTPUT = (
    LOG_FOLDER
    / "model_eda_spearman_correlation.csv"
)


TARGET_CORRELATION_OUTPUT = (
    LOG_FOLDER
    / "model_eda_target_correlation.csv"
)


HIGH_CORRELATION_OUTPUT = (
    LOG_FOLDER
    / "model_eda_high_correlation_pairs.csv"
)


VIF_OUTPUT = (
    LOG_FOLDER
    / "model_eda_vif.csv"
)


STATION_SUMMARY_OUTPUT = (
    LOG_FOLDER
    / "model_eda_station_summary.csv"
)


MONTHLY_SUMMARY_OUTPUT = (
    LOG_FOLDER
    / "model_eda_monthly_summary.csv"
)


SEASONAL_SUMMARY_OUTPUT = (
    LOG_FOLDER
    / "model_eda_seasonal_summary.csv"
)


ANNUAL_SUMMARY_OUTPUT = (
    LOG_FOLDER
    / "model_eda_annual_summary.csv"
)


# ======================================================================
# 26. SAVE OUTPUTS
# ======================================================================

eda_df.to_csv(
    EDA_READY_OUTPUT,
    index=False
)


outlier_filtered_df.to_csv(
    OUTLIER_FILTERED_OUTPUT,
    index=False
)


overall_summary_df.to_csv(
    OVERALL_SUMMARY_OUTPUT,
    index=False
)


missing_summary_df.to_csv(
    MISSING_SUMMARY_OUTPUT,
    index=False
)


numeric_summary_df.to_csv(
    NUMERIC_SUMMARY_OUTPUT,
    index=False
)


predictor_quality_df.to_csv(
    PREDICTOR_QUALITY_OUTPUT,
    index=False
)


target_summary_df.to_csv(
    TARGET_SUMMARY_OUTPUT,
    index=False
)


outlier_summary_df.to_csv(
    OUTLIER_SUMMARY_OUTPUT,
    index=False
)


pearson_correlation_df.to_csv(
    PEARSON_CORRELATION_OUTPUT
)


spearman_correlation_df.to_csv(
    SPEARMAN_CORRELATION_OUTPUT
)


target_correlation_df.to_csv(
    TARGET_CORRELATION_OUTPUT,
    index=False
)


high_correlation_pairs_df.to_csv(
    HIGH_CORRELATION_OUTPUT,
    index=False
)


vif_df.to_csv(
    VIF_OUTPUT,
    index=False
)


station_summary_df.to_csv(
    STATION_SUMMARY_OUTPUT,
    index=False
)


monthly_summary_df.to_csv(
    MONTHLY_SUMMARY_OUTPUT,
    index=False
)


seasonal_summary_df.to_csv(
    SEASONAL_SUMMARY_OUTPUT,
    index=False
)


annual_summary_df.to_csv(
    ANNUAL_SUMMARY_OUTPUT,
    index=False
)


# ======================================================================
# 27. RECOMMENDED FEATURE GROUPS
# ======================================================================

constant_features = (
    predictor_quality_df[
        predictor_quality_df[
            "Constant"
        ]
    ][
        "Predictor"
    ]
    .tolist()
)


near_constant_features = (
    predictor_quality_df[
        predictor_quality_df[
            "Near_Constant"
        ]
    ][
        "Predictor"
    ]
    .tolist()
)


high_vif_features = (
    vif_df[
        vif_df[
            "High_VIF"
        ]
    ][
        "Feature"
    ]
    .tolist()
    if not vif_df.empty
    else []
)


recommended_base_features = [
    feature
    for feature in (
        available_predictors
        + TEMPORAL_FEATURES
        + COORDINATE_FEATURES
    )
    if feature in eda_df.columns
    and feature not in constant_features
]


recommended_without_gsmap_gauge = [
    feature
    for feature in recommended_base_features
    if feature != "GSMaP_Gauge"
]


recommended_without_static_features = [
    feature
    for feature in recommended_base_features
    if feature not in [
        "Distance_Sea_meter",
        "Land_Aspect_Degree_30m",
        "Land_Khulna_SRTM_DEM",
        "Land_Slope_Degree_30m",
    ]
]


print()
print("=" * 82)
print("FEATURE SCREENING SUMMARY")
print("=" * 82)

print(
    f"Constant features      : "
    f"{constant_features}"
)

print(
    f"Near-constant features : "
    f"{near_constant_features}"
)

print(
    f"High-VIF features      : "
    f"{high_vif_features}"
)

print()
print("Recommended base feature list:")

for number, feature in enumerate(
    recommended_base_features,
    start=1
):

    print(
        f"{number:02d}. {feature}"
    )


# ======================================================================
# 28. FINAL COMPLETION MESSAGE
# ======================================================================

print()
print("=" * 82)
print("MODEL DATA EXPLORATION COMPLETED")
print("=" * 82)

print(
    f"Input rows                   : "
    f"{len(df)}"
)

print(
    f"EDA-ready rows               : "
    f"{len(eda_df)}"
)

print(
    f"Extreme-outlier-filtered rows: "
    f"{len(outlier_filtered_df)}"
)

print(
    f"Available predictors         : "
    f"{len(available_predictors)}"
)

print(
    f"Duplicate rows               : "
    f"{duplicate_count}"
)

print(
    f"Missing target rows          : "
    f"{missing_target_count}"
)

print(
    f"Target IQR outliers          : "
    f"{target_outlier_count}"
)

print(
    f"Target extreme outliers      : "
    f"{target_extreme_outlier_count}"
)

print(
    f"High-correlation pairs       : "
    f"{len(high_correlation_pairs_df)}"
)

print(
    f"High-VIF features            : "
    f"{len(high_vif_features)}"
)

print(
    f"Figures created              : "
    f"{len(created_figures)}"
)


print()
print("Main output datasets:")

print(
    f"1. {EDA_READY_OUTPUT}"
)

print(
    f"2. {OUTLIER_FILTERED_OUTPUT}"
)


print()
print("Important logs:")

print(
    f"1. {OVERALL_SUMMARY_OUTPUT}"
)

print(
    f"2. {PREDICTOR_QUALITY_OUTPUT}"
)

print(
    f"3. {TARGET_CORRELATION_OUTPUT}"
)

print(
    f"4. {HIGH_CORRELATION_OUTPUT}"
)

print(
    f"5. {VIF_OUTPUT}"
)

print(
    f"6. {OUTLIER_SUMMARY_OUTPUT}"
)


print()
print("Figures saved in:")

print(
    FIGURE_FOLDER
)
```
